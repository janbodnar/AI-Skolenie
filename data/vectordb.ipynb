```python
import chromadb

# 1. Initialize the Chroma client (runs entirely in-memory for this demo)
client = chromadb.Client()

# 2. Create a collection (a namespace for your vectors)
collection = client.get_or_create_collection(name="ai_course_demo2")

# 3. Add documents (Chroma can automatically generate embeddings using default models)
collection.add(
    documents=[
        "Transformers use self-attention mechanisms to process text in parallel.",
        "Vector databases store high-dimensional embeddings for fast similarity search.",
        "Gradient descent is an optimization algorithm used to train neural networks.",
        "A pizza is a savory dish of Italian origin consisting of a round, flattened base."
    ],
    metadatas=[{"topic": "NLP"}, {"topic": "Databases"}, {"topic": "Training"}, {"topic": "Food"}],
    ids=["doc1", "doc2", "doc3", "doc4"]
)

# 4. Query the database semantically (notice we aren't using keyword matching)
results = collection.query(
    query_texts=["What Origin has pizza?"],
    where={"topic": "Food"},
    n_results=2 # Return top 2 closest matches
)

print("Query Results:")
for doc, meta in zip(results['documents'][0], results['metadatas'][0]):
    print(f"- [{meta['topic']}] {doc}")
```

```python
# Fetch all stored documents and their vector embeddings
all_data = collection.get(include=["embeddings", "documents", "metadatas"])

print(f"Total documents in collection: {len(all_data['ids'])}\n")

for doc_id, text, vector in zip(all_data['ids'], all_data['documents'], all_data['embeddings']):
    print(f"ID: {doc_id}")
    print(f"Text: '{text}'")
    print(f"Vector Dimensions: {len(vector)}")
    print(f"First 5 numbers of vector: {vector[:5]}...\n" + "-"*50)
```
