```python
!pip install -q tiktoken
import tiktoken
enc = tiktoken.get_encoding("o200k_base")
for t in ["Good morning, how are you?", "Dobré ráno, ako sa máš?"]:
    ids = enc.encode(t)
    print(len(ids), [enc.decode([i]) for i in ids])
```

```python
words = ["strawberry", "strawberries", "anthropomorphism"]
for w in words:
    ids = enc.encode(w)
    print(f"'{w}' -> {len(ids)} tokens: {[enc.decode([i]) for i in ids]}")
```

```python
numbers = ["12", "123", "1234", "12345", "123456", "3891029"]
for n in numbers:
    ids = enc.encode(n)
    print(f"'{n}' -> {[enc.decode([i]) for i in ids]}")
```

```python
variations = ["hello", " hello", "Hello", "HELLO"]
for v in variations:
    ids = enc.encode(v)
    print(f"'{v}' -> IDs: {ids} | Tokens: {[enc.decode([i]) for i in ids]}")
```

```python
phrases = [
    "Hello, how are you?",           # English
    "Ahoj, ako sa máš?",             # Slovak
    "こんにちは、お元気ですか？"        # Japanese
]
for p in phrases:
    ids = enc.encode(p)
    print(f"{len(ids)} tokens: {[enc.decode([i]) for i in ids]}")
```
